```python
spark.sql("""
CREATE OR REPLACE TABLE proyecto_bi.gold.fact_viajes AS
SELECT
    VendorID                                            AS vendor_id,
    DATE(tpep_pickup_datetime)                          AS fecha,
    HOUR(tpep_pickup_datetime)                          AS hora,
    PULocationID                                        AS pickup_location_id,
    DOLocationID                                        AS dropoff_location_id,
    payment_type                                        AS payment_type_id,
    passenger_count,
    trip_distance,
    fare_amount,
    tip_amount,
    tolls_amount,
    total_amount,
    congestion_surcharge,
    DATEDIFF(MINUTE, tpep_pickup_datetime, tpep_dropoff_datetime) AS duracion_minutos
FROM proyecto_bi.silver.yellow_trips
""")
print("✅ fact_viajes creada")
```

```python
%sql
SELECT COUNT(*) AS total_viajes FROM proyecto_bi.gold.fact_viajes;
```
