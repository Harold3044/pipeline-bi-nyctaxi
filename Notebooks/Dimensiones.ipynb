```python
# DIMENSIÓN TIEMPO
spark.sql("""
CREATE OR REPLACE TABLE proyecto_bi.gold.dim_tiempo AS
SELECT DISTINCT
    DATE(tpep_pickup_datetime)                          AS fecha,
    YEAR(tpep_pickup_datetime)                          AS anio,
    MONTH(tpep_pickup_datetime)                         AS mes,
    DAY(tpep_pickup_datetime)                           AS dia,
    HOUR(tpep_pickup_datetime)                          AS hora,
    DAYOFWEEK(tpep_pickup_datetime)                     AS dia_semana,
    CASE DAYOFWEEK(tpep_pickup_datetime)
        WHEN 1 THEN 'Domingo'
        WHEN 2 THEN 'Lunes'
        WHEN 3 THEN 'Martes'
        WHEN 4 THEN 'Miércoles'
        WHEN 5 THEN 'Jueves'
        WHEN 6 THEN 'Viernes'
        WHEN 7 THEN 'Sábado'
    END                                                 AS nombre_dia,
    CASE 
        WHEN DAYOFWEEK(tpep_pickup_datetime) IN (1,7) 
        THEN 'Fin de semana' 
        ELSE 'Día hábil' 
    END                                                 AS tipo_dia
FROM proyecto_bi.silver.yellow_trips
""")
print("✅ dim_tiempo creada")

# DIMENSIÓN PAGO
spark.sql("""
CREATE OR REPLACE TABLE proyecto_bi.gold.dim_pago AS
SELECT DISTINCT
    payment_type                                        AS payment_type_id,
    CASE payment_type
        WHEN 1 THEN 'Tarjeta de crédito'
        WHEN 2 THEN 'Efectivo'
        WHEN 3 THEN 'Sin cargo'
        WHEN 4 THEN 'Disputa'
        WHEN 5 THEN 'Desconocido'
        WHEN 6 THEN 'Viaje anulado'
    END                                                 AS metodo_pago
FROM proyecto_bi.silver.yellow_trips
""")
print("✅ dim_pago creada")

# DIMENSIÓN VENDOR
spark.sql("""
CREATE OR REPLACE TABLE proyecto_bi.gold.dim_vendor AS
SELECT DISTINCT
    VendorID                                            AS vendor_id,
    CASE VendorID
        WHEN 1 THEN 'Creative Mobile Technologies'
        WHEN 2 THEN 'VeriFone Inc'
    END                                                 AS nombre_vendor
FROM proyecto_bi.silver.yellow_trips
""")
print("✅ dim_vendor creada")
```
